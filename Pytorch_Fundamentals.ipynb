### reshaping , stacking , squeezing and unsqueezing
* reshaping - resahpe the tensors shape
* view - return view of an input tensors of certain shape but same memory as original tensor
* stacking - combing multiple tensors on top eachother(vstack) or sideby side(hastack)
* squeeze- removing all 1 dimensions from tensors
* unsqueeze - add 1 dim to target tensor
* permute - return a view of input with dim permuted in certain way

```python
# lets create tensors
import tensor
x = x.arange(1., 10.)   # 9 values  (3,3) (1,9) (9,1)......
x
#add extra dimension
x_shaped = x.reshape(9,1)

# change the view
z = x.view(1,9)
z, z.shape            #same as before result

# changing z changes x
z[:,0] = 5
z, x                  # if z values changes means x also changed

```

```python
# stacks tensors on top
x_stacked = torch.stack([x,x,x,x], dim = 0]    # 0 means horizontal we get 4 rows 4*9
                                                 # 1 means vetical we get 9*4 2 means you got error


# sequeeze and unsqueeze
x.                                # ([1,9])
x.squeeze()                     # reamove all single dimensions   [1,9]   ==> ([9])

x.unsqueeze(dim = 1)        # add 1 dim ([9]) ==> ([9,1])   1 means vertical

# permute   means change the order of dim
x = torch.randn(2,3,5)
x.size()                # ([2,3,5])

torch.permute(x, (2,0,1)).size()         # ([5,2,3])          but view is original not change
```
